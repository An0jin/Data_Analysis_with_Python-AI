# 남녀 인구수 데이터 분리

```python
import pandas as pd

# 1. 데이터 불러오기
df = pd.read_csv('인구수.csv')

# 2. 남자 인구수 데이터만 추출 후 저장
cols_male = ['행정구역'] + [c for c in df.columns if '남자' in c]
df_male = df[cols_male]
df_male.to_csv('남자_인구수.csv', index=False)

# 3. 여자 인구수 데이터만 추출 후 저장
cols_female = ['행정구역'] + [c for c in df.columns if '여자' in c]
df_female = df[cols_female]
df_female.to_csv('여자_인구수.csv', index=False)

print("남자_인구수.csv, 여자_인구수.csv 파일 생성 완료!")
```
